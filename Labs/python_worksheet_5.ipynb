## First Plot

1. Plot cos x for x from 0 to 2 pi using 100 points. Save it as cos.png.

```python
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
matplotlib.use("Agg")

x = np.linspace(0, 2*np.pi, 100)
y = np.cosh(x)

plt.plot(x,y)
plt.savefig("cos.png")
plt.close()
```

2. Plot y = x squared for x from -5 to 5 using 50 points. Save it as parabola.png.

```python
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
matplotlib.use("Agg")

x = np.linspace(-5, 5, 50)
y = x**2

plt.plot(x,y)
plt.savefig("parabola.png")
plt.close()
```

3. Plot y = exp(-x) for x from 0 to 5. Save it as decay.png.

```python
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
matplotlib.use("Agg")

x = np.linspace(0, 5, 100)
y = np.exp(-x)

plt.plot(x,y)
plt.savefig("decay.png")
plt.close()

```

## Labels, title and legend

4. Plot sin x and cos x on the same axes for x from 0 to 2 pi. Add label=
, x and y labels, a
title, a legend, and save with dpi=150.

```python
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
matplotlib.use("Agg")

x = np.linspace(0, 2*np.pi, 100)

plt.plot(x, np.sin(x), label = "sin x")
plt.plot(x, np.cos(x), label = "cos x")
plt.xlabel("x")
plt.ylabel("y")
plt.title("sine and cosine function")
plt.legend()
plt.savefig("sincos.png", dpi = 150)
plt.close()
```

5. Plot x, x**2 and x**3 for x from 0 to 2 on the same axes, with a legend.

```python
import matplotlib
import numpy as np
import matplotlib.pyplot as plt
matplotlib.use("Agg")

x = np.linspace(0, 2, 200)

plt.plot(x, x, label = "x")
plt.plot(x, x**2, label = "x2")
plt.plot(x, x**3, label = "x3")
plt.xlabel("x")
plt.ylabel("x to the power")
plt.title("x's power")
plt.legend()
plt.savefig("to_the_power.png")
plt.close()
```
