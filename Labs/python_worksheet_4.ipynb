### A list is not a vector

4. Rewrite this loop as one array expression: build a list of the squares of 1 to 5 by looping
and appending.

```python
sq = []
for i in range(1,6):
    sq.append(i**2)
print(sq)

sq2 = np.array(range(1,6))**2
print(sq2)
```
