Question 4

```python
x = input("Enter an english alphabet: ")
vowels = ['a', 'e', 'i', 'o', 'u']
consonants = ['b', 'c', 'd', 'f', 'g', 'h', 'j', 'k', 'l', 'm', 'n', 'p', 'q', 'r', 's', 't', 'v', 'w', 'x', 'y', 'z']
if x in vowels:
    print("Vowel")
elif x in consonants:
    print("Consonant")
else:
    print("Invalid input")
```
