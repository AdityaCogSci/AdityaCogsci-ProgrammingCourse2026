**Question 2.** A ball is thrown straight up at u = 20 m/s. Its height is h = u t − ½ g t². Rearranged, (g/2) t² − u t + h = 0.

For a target height h, compute the discriminant and print one of three messages: the ball reaches h twice (print both times), just touches h at the top, or never reaches h (print the maximum height u²/2g).

Test with h = 15 m and h = 25 m.

```python
# Question 2
import math

u = 20.0
g = 9.81
h_max = u**2 / (2 * g)

for h in [15, 25]:
    a = 0.5 * g
    b = -u
    c = h
    disc = b**2 - 4 * a * c
    
    print(f"Testing for h = {h} m (discriminant = {disc:.2f}):")
    if disc > 0:
        t1 = (-b - math.sqrt(disc)) / (2 * a)
        t2 = (-b + math.sqrt(disc)) / (2 * a)
        print(f"Ball reaches h twice: at t1 = {t1:.2f} s and t2 = {t2:.2f} s")
    elif disc == 0:
        t = -b / (2 * a)
        print(f"Ball just touches h at the top at t = {t:.2f} s")
    else:
        print(f"Ball never reaches h. Maximum height is {h_max:.2f} m")
    print()
```

**Question 13.** Write factorial(n) using a for loop. Then write exp_series(x, nterms=10) that adds the first nterms terms of Σ xⁿ / n!, calling factorial inside it.

For x = 2, print the approximation and its error compared with math.exp(2) for nterms = 2, 4, 8 and 16.

```python
# Question 13
import math

def factorial(n):
    res = 1
    for i in range(1, n + 1):
        res *= i
    return res

def exp_series(x, nterms=10):
    total = 0.0
    for n in range(nterms):
        total += (x**n) / factorial(n)
    return total

x = 2
exact_val = math.exp(x)
print(f"math.exp({x}) = {exact_val:.10f}\n")

for nterms in [2, 4, 8, 16]:
    approx = exp_series(x, nterms=nterms)
    err = abs(exact_val - approx)
    print(f"nterms = {nterms:2d}: approx = {approx:.10f}, error = {err:.10e}")
```

```python
# Question 14
def newton_sqrt(a, tol=1e-10):
    x = float(a)
    iters = 0
    while abs(x**2 - a) > tol:
        x = 0.5 * (x + a / x)
        iters += 1
    return x, iters

tests = [(2, 1e-10), (9, 1e-10), (10**6, 1e-10), (2, 1e-3)]
for a, tol in tests:
    val, iters = newton_sqrt(a, tol=tol)
    print(f"a = {a:<7}, tol = {tol:1.0e} -> sqrt = {val:.10f}, iterations = {iters}")
```

```python
# Question 29
import numpy as np

V = [1, 2, 3, 4, 5]
I = [0.21, 0.39, 0.62, 0.80, 1.01]

with open("ohm.csv", "w") as f:
    f.write("V,I\n")
    for v_val, i_val in zip(V, I):
        f.write(f"{v_val},{i_val}\n")

data = np.loadtxt("ohm.csv", delimiter=",", skiprows=1)
V_loaded = data[:, 0]
I_loaded = data[:, 1]
R = V_loaded / I_loaded

print("Resistance for each measurement:")
for i in range(len(R)):
    print(f"  V = {V_loaded[i]:.0f} V, I = {I_loaded[i]:.2f} A -> R = {R[i]:.2f} ohms")

print(f"\nMean resistance: {np.mean(R):.2f} ohms")
```

**Question 30.** Make t = np.linspace(0, 2, 201) and x = cos(2πt).

Save x with np.save, load it back, and print the shape and the largest difference from the original.

Save t and x together with np.savez("run1.npz", time=t, position=x). Load the file and print the first three times and the last position.

```python
# Question 30
import numpy as np

t = np.linspace(0, 2, 201)
x = np.cos(2 * np.pi * t)

np.save("position.npy", x)
x_loaded = np.load("position.npy")

print("Loaded position shape:", x_loaded.shape)
print("Max difference from original:", np.max(np.abs(x - x_loaded)))

np.savez("run1.npz", time=t, position=x)
data_npz = np.load("run1.npz")

print("First three times:", data_npz["time"][:3])
print("Last position:    ", data_npz["position"][-1])
```
